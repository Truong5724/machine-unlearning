```python
import os

times_dir = 'containers/utkface_ovr/times'

# Check if the directory exists
if not os.path.exists(times_dir):
    print(f"Thư mục '{times_dir}' không tồn tại. Vui lòng kiểm tra lại đường dẫn.")
else:
    total_train_time = 0.0
    # Modified filter to include 'shard-X_0.time' files
    time_files = [f for f in os.listdir(times_dir) if f.startswith('shard-') and f.endswith('.time')]

    if not time_files:
        print(f"Không tìm thấy file thời gian (.time) nào trong thư mục '{times_dir}'.")
    else:
        print("Thời gian train từng shard:")
        for filename in sorted(time_files):
            filepath = os.path.join(times_dir, filename)
            try:
                with open(filepath, 'r') as f:
                    # Assuming each .time file contains a single float value
                    time_taken = float(f.read().strip())
                    print(f"  {filename}: {time_taken:.2f} giây")
                    total_train_time += time_taken
            except ValueError:
                print(f"  Lỗi đọc file '{filename}': Nội dung không phải là số hợp lệ.")
            except Exception as e:
                print(f"  Lỗi khi xử lý file '{filename}': {e}")

        print(f"\nTổng thời gian train cho tất cả các shard là: {total_train_time:.2f} giây")
        print(f"Tương đương: {total_train_time / 3600:.2f} giờ")
```
