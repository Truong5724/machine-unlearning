## Evaluate Single Unlearned Model (Shard 0) on Forgotten Data

### Subtask:
Evaluate the accuracy of the specific unlearned model from shard 0 (containers/utkface/cache/shard-0:100.pt) on the data identified as forgotten by shard 0 (from requestfile:100.npy).


**Reasoning**:
The subtask requires creating a Python script to evaluate a single unlearned model on its forgotten data. I will use the `%%writefile` magic command to create the `evaluate_single_shard.py` file with the provided code.



```python
%%writefile evaluate_single_shard.py
import numpy as np
import torch
import torch.nn as nn
import os
from architectures import utkface as architectures
from datasets.UTKFace.dataloader import X_train, y_train # Corrected import

# Load request file for unlearned data (epoch 100)
request_file_path = "containers/utkface/requestfile:100.npy"
requests = np.load(request_file_path, allow_pickle=True)

# Get the indices of data points forgotten by shard 0
shard_0_forgotten_indices = requests[0]

# Load the unlearned model for shard 0 (epoch 100)
model_path = "containers/utkface/cache/shard-0:100.pt"
# Corrected model instantiation: use 'Model' class and pass required arguments
model = architectures.Model(input_shape=(3, 64, 64), nb_classes=2)
# Map to CPU for inference if a GPU isn't available or if the model was saved on GPU
model.load_state_dict(torch.load(model_path, map_location=torch.device('cpu')))
model.eval()

# Prepare data for prediction
# Select only the forgotten data points for shard 0 from the training set
x_forgotten_shard_0 = torch.from_numpy(X_train[shard_0_forgotten_indices]).float()
y_forgotten_shard_0 = y_train[shard_0_forgotten_indices]

# Make predictions
predictions = []
with torch.no_grad():
    outputs = model(x_forgotten_shard_0)
    _, predicted = torch.max(outputs.data, 1)
    predictions = predicted.numpy()

# Calculate accuracy
correct_predictions = (predictions == y_forgotten_shard_0).sum().item()
total_predictions = len(y_forgotten_shard_0)
accuracy = correct_predictions / total_predictions if total_predictions > 0 else 0

print(f"Accuracy of unlearned model from shard 0 on its forgotten data: {accuracy:.4f}")
```

## Evaluate All Unlearned Models on Forgotten Data

### Subtask:
Evaluate the combined accuracy of all four unlearned models (from epoch 100) on the data identified as forgotten in `requestfile:100.npy`.


```python
%%writefile evaluate_all_shards.py
import numpy as np
import torch
import torch.nn as nn
import os
from architectures import utkface as architectures
from datasets.UTKFace.dataloader import X_train, y_train # Corrected import

# Define the number of shards
NUM_SHARDS = 4

# Load request file for unlearned data (epoch 100)
request_file_path = "containers/utkface/requestfile:100.npy"
requests = np.load(request_file_path, allow_pickle=True)

# Initialize lists to store all forgotten data points' actual labels and predictions
all_true_labels = []
all_predictions = []

print(f"Evaluating combined accuracy of {NUM_SHARDS} models on their forgotten data (shard-0 unlearned, others retained)...")

# Loop through each shard
for i in range(NUM_SHARDS):
    print(f"Processing shard {i}...")

    # Determine the correct model path based on the shard index
    if i == 0:
        # Load the unlearned model for shard 0 (epoch 100)
        current_model_path = f"containers/utkface/cache/shard-{i}:100.pt"
    else:
        # Load the retained/baseline model for other shards (epoch 0, meaning full training)
        current_model_path = f"containers/utkface/cache/shard-{i}:0.pt"

    model = architectures.Model(input_shape=(3, 64, 64), nb_classes=2)
    model.load_state_dict(torch.load(current_model_path, map_location=torch.device('cpu')))
    model.eval()

    # Get the indices of data points forgotten by the current shard
    shard_forgotten_indices = requests[i]

    # If there are forgotten indices for this shard, make predictions
    if len(shard_forgotten_indices) > 0:
        # Select only the forgotten data points for the current shard from the training set
        x_forgotten = torch.from_numpy(X_train[shard_forgotten_indices]).float()
        y_forgotten = y_train[shard_forgotten_indices]

        # Make predictions
        with torch.no_grad():
            outputs = model(x_forgotten)
            _, predicted = torch.max(outputs.data, 1)

        all_true_labels.extend(y_forgotten.tolist())
        all_predictions.extend(predicted.numpy().tolist())
    else:
        print(f"No forgotten data for shard {i}.")

# Convert lists to numpy arrays for calculation
all_true_labels = np.array(all_true_labels)
all_predictions = np.array(all_predictions)

# Calculate combined accuracy
if len(all_true_labels) > 0:
    correct_predictions = (all_predictions == all_true_labels).sum().item()
    total_predictions = len(all_true_labels)
    combined_accuracy = correct_predictions / total_predictions
    print(f"\nCombined accuracy of all models on forgotten data (shard-0 unlearned, others retained): {combined_accuracy:.4f}")
else:
    print("No forgotten data across all shards to evaluate.")
```
