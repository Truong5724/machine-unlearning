```python
import pandas as pd

# Read the original attribute file, which is now named list_attr_celeba.txt
df_attr = pd.read_csv('list_attr_celeba.csv')

# Get the number of images
num_images = len(df_attr)

# Get attribute names (excluding 'image_id') and format them as space-separated
attr_names = ' '.join(df_attr.columns[1:])

# Prepare the data lines: image_id followed by space-separated attribute values
data_lines = []
for index, row in df_attr.iterrows():
    image_id = row['image_id']
    attributes = ' '.join(map(str, row[1:].values))
    data_lines.append(f"{image_id} {attributes}")

# Combine all parts into the expected format
formatted_content = f"{num_images}\n{attr_names}\n" + '\n'.join(data_lines)

# Overwrite the list_attr_celeba.txt file with the correctly formatted content
with open('list_attr_celeba.txt', 'w') as f:
    f.write(formatted_content)

print("Successfully reformatted list_attr_celeba.txt")
```
